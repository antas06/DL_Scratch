```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
#Loading the dataset
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
print(y_train.shape)
print(X_train.shape)
print(y_test.shape)
print(X_test.shape)
```

Thus, total 60000 samples of data, each with 28 * 28 pikel image, thus a total of 28*28 inputs, and 10 outputs.

```python
X_train[0]
```

```python
#Normalising
X_train = X_train/255
X_test = X_test/255
```

```python
X_train[0][7]  #Random Row
```

```python
model = Sequential()

model.add(Flatten(input_shape=(28,28)))  #to 1-D Array
model.add(Dense(150, activation="relu"))
model.add(Dense(30,activation="relu"))
model.add(Dense(10,activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=["accuracy"])
```

```python
model_epo15 = model.fit(X_train,y_train,epochs=15, validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
```

```python
y_pred = y_prob.argmax(axis=1)
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
for i in range(100):
  if y_pred[i] == y_test[i]:
    print(y_test[i],"is truely predicted as", y_pred[i])
  else:
    print(y_test[i],"is predicted as", y_pred[i])
```
