```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Embedding, SimpleRNN, Dense
```

```python
#taking an example of a data set
#30 sentences - first 15 positive and next 15 negative
sentences = [
 "I love this product",
 "This movie made me smile",
 "Service was friendly and quick",
 "Today felt bright and happy",
 "This is the best day",
 "Absolutely fantastic experience",
 "I enjoyed every single moment",
 "Great job, well done",
 "The food tasted delicious",
 "Totally recommend to everyone",
 "Very satisfied with results",
 "This worked better than expected",
 "Amazing quality and value",
 "Such a pleasant surprise",
 "I feel positive about this",
 "I hate this product",
 "This movie bored me",
 "Service was rude and slow",
 "Today was cold and lonely",
 "This is the worst day",
 "Terrible experience overall",
 "I regret buying this",
 "Very disappointed with results",
 "The food tasted awful",
 "Do not recommend this",
 "It broke after one use",
 "Not worth the money",
 "Utterly frustrating and annoying",
 "I feel negative about this",
 "Such a waste of time",
]
labels = [1]*15 + [0]*15
labels = np.array(labels)
```

```python
labels
```

```python
#Tokenising all words
vocab_size = 2000
toki  = Tokenizer(num_words = vocab_size,oov_token = "") #oov-> Out of vocab
toki.fit_on_texts(sentences) #assigning the words an index(this is how tokenizer work in kearas)

seqs = toki.texts_to_sequences(sentences) #sequencing the whole sentance eg. [2, 5, 7]
maxlen = max(len(s) for s in seqs)
X = pad_sequences(seqs,maxlen = maxlen,padding = 'post') #post- padding to the maximum sequence length
y = labels
#ready with our X and y
```

```python
#defining the embedding dimentions and rnn unit
embed_dim = 16
rnn_units = 8
```

```python
#defining the Input (Kearas Layer)
inp = Input(shape= (maxlen,),dtype="int32",name = 'input')

#inp kayega isme - Embedding layer(kearas layer)
x = Embedding(input_dim=vocab_size,output_dim=embed_dim,mask_zero=True,name = 'embed')(inp)

#enn layer (kearas SimpleRNN)
rnn_x = SimpleRNN(units = rnn_units, return_sequences = False,return_state= False,name = 'simple_rnn')(x)

#the output layer(Kearas Dense)
out = Dense(1,activation='sigmoid',name = 'out')(rnn_x)

#final model bna diya and compiling
model = Model(inputs =inp,outputs = out)
model.compile(optimizer='adam',loss = 'binary_crossentropy',metrics=['accuracy'])
model.summary()
```

```python
model.fit(X,y, epochs= 25,batch_size = 8,verbose = 1)
```

```python
sente = ["Amazing Product", "Worst Product"]
seq = toki.texts_to_sequences(sente)
padi = pad_sequences(seq,maxlen = maxlen,padding = 'post')
pred = model.predict(padi)
print(pred)
```
