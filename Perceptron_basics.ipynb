```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#uploaded a dtatset for the perceptron training
file = pd.read_csv("/content/perceptron_dataset_100.csv")
```

```python
X = file.iloc[:,:-1].values
y = file.iloc[:,-1].values
```

```python
def perceptron(X,y):
    w1=w2=b=1
    lr = 0.1
    for j in range(1000):
      for i in range(X.shape[0]):
            # yeh hai f(x)
            z = w1*X[i][0] + w2*X[i][1] + b

            #Agar result opposite h
            if z*y[i] < 0:
                w1 = w1 + lr*y[i]*X[i][0]
                w2 = w2 + lr*y[i]*X[i][1]
                b = b + lr*y[i]

    return w1,w2,b
```

```python
w1,w2,b = perceptron(X,y)
```

```python
m = -(w1/w2)
c = -(b/w2)
```

```python
x_input = np.linspace(-4,4,100)
y_input = m * x_input + c
y_input
```

```python
plt.scatter(file["x1"], file["x2"], c=file["label"])
plt.plot(x_input, y_input, color='red')
plt.show()
```
