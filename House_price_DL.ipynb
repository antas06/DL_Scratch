```python
import pandas as pd

# Create the dataset as a pandas DataFrame from the provided sample data
data = {
    'Area (sq.ft.)': [1000, 1200, 1500, 1800, 2000],
    'Bedrooms': [2, 2, 3, 3, 4],
    'Age (years)': [10, 5, 8, 5, 3],
    'Price (₹ lakh)': [45, 55, 70, 85, 100]
}
df = pd.DataFrame(data)

# Display the DataFrame to verify
display(df)
```

```python
# Separate features (X) and target (y)
X = df[['Area (sq.ft.)', 'Bedrooms', 'Age (years)']]
y = df['Price (₹ lakh)']

# Display the first few rows of X and y to verify
print('Features (X):')
display(X.head())
print('\nTarget (y):')
display(y.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

# Initialize the MinMaxScaler
scaler = MinMaxScaler()

# Fit the scaler to the features and transform them
X_scaled = scaler.fit_transform(X)

# Convert the scaled features back to a DataFrame for better readability and to maintain column names
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# Display the first few rows of the scaled features
print('Scaled Features (X_scaled):')
display(X_scaled_df.head())
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
# We'll use 80% for training and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Display the shapes of the resulting datasets to verify the split
print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

# Define the Deep Neural Network model
model = keras.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)), # Input layer with 64 neurons
    layers.Dense(64, activation='relu'), # Hidden layer with 64 neurons
    layers.Dense(1)  # Output layer with 1 neuron for regression (predicting price)
])

# Compile the model
# Using Adam optimizer and Mean Squared Error (MSE) as the loss function
model.compile(optimizer='adam', loss='mse', metrics=['mae', 'mse'])

# Display the model summary
model.summary()
```

```python
# Train the model
history = model.fit(
    X_train,
    y_train,
    epochs=100,  # Number of training epochs
    batch_size=1, # Small dataset, so batch_size can be 1 or adjusted
    verbose=0 # Suppress output for each epoch
)

print("Model training complete.")
```

```python
# Evaluate the model on the test data
loss, mae, mse = model.evaluate(X_test, y_test, verbose=0)

print(f"\nModel Evaluation on Test Data:")
print(f"Loss (MSE): {loss:.2f}")
print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
```
