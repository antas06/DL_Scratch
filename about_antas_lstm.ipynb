```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer(oov_token="<OOV>")
tokenizer.fit_on_texts(corpus)

print(tokenizer.word_index)
print(len(tokenizer.word_index))
```

```python
Sentes = []

for esen in corpus:
  tokenised_sente = tokenizer.texts_to_sequences([esen])[0]
  Sentes.append(tokenised_sente)

print(len(Sentes))
```

```python
f_td= []

for i in range (len(Sentes)):
  for l in range(1, len(Sentes[i])):
    f_td.append(Sentes[i][:l+1])


print(f_td)
print(len(f_td))
```

```python
max_len = max([len(x) for x in f_td])
print(max_len)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
p_td = pad_sequences(f_td, maxlen = max_len, padding='pre')

print(p_td)
```

```python
X = p_td[:, :-1]
y = p_td[:, -1]

print(X.shape)
print(y.shape)
```

```python
y
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python
model = Sequential()
vocab_size = len(tokenizer.word_index) + 1

model.add(Embedding(vocab_size, 100, input_length=13))
model.add(LSTM(150,return_sequences=True))
model.add(LSTM(150))
model.add(Dense(vocab_size, activation='softmax'))
```

```python
model.compile(loss='categorical_crossentropy', optimizer='adam',metrics=['accuracy'])
model.summary()
```

```python
y = tf.keras.utils.to_categorical(y, num_classes=vocab_size) #ohe
model.fit(X,y,epochs=120)
```

```python
text = "antas"

for i in range(10):
  token_text = tokenizer.texts_to_sequences([text])[0]
  padded_token_text = pad_sequences([token_text], maxlen=56, padding='pre')
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
```
